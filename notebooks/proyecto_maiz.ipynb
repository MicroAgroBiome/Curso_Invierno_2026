**Pregunta del proyecto:** dos muestras de rizósfera de maíz, ¿se parecen más si vienen del **mismo estudio**, de **otro estudio del mismo país**, o de **otro país**?

**Datos:** 41 metagenomas de rizósfera de maíz (*Zea mays*) de la base de datos [MicroAgroBiome](https://agrobiom.matmor.unam.mx/), con conteos de lecturas por género:

- 31 muestras de **Sudáfrica** (5 estudios)
- 10 muestras de **Estados Unidos** (1 estudio)

⚠️ **Limitación del diseño:** las muestras de Estados Unidos vienen de un solo estudio, así que para ese país *país* y *estudio* son la misma variable. Lo que sí pueden medir es cuánto varía el maíz **entre estudios dentro de Sudáfrica**.
